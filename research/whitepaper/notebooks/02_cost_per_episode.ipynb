# Per-episode cost — Tier H/M/L decomposition


_12-stage cost breakdown for default 60s episode at scene_reuse=0.40, retry_factor=0.18._

> Reproducible from `data/computed/*.json` written by `run_all.py`.


```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
cost = load('cost')
for tier in ('H', 'M', 'L'):
    print(tier, 'mean =', cost[f'tier_{tier}']['mc_mean'], 'p95 =', cost[f'tier_{tier}']['mc_p95'])
stages = cost['tier_M']['stages']
names = [s['name'] for s in stages]
vals = [s['with_retry_cny'] for s in stages]
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.barh(names, vals); ax.axvline(80, color='r', linestyle='--', label='need.md ≤ ¥80')
ax.set_xlabel('CNY/episode'); ax.set_title('Tier M cost decomposition'); ax.legend(); plt.tight_layout(); plt.show()
```
