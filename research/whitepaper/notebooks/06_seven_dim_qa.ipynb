# 7-dim QA pass-rate (Beta MC)


_Per-dimension and joint pass rate at threshold ∈ {7.0,7.5,8.0,8.5,9.0}._

> Reproducible from `data/computed/*.json` written by `run_all.py`.


```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
sd = load('seven_dim_qa')
thrs = [7.0,7.5,8.0,8.5,9.0]; rates = [sd[f'threshold_{t}']['pass_rate'] for t in thrs]
plt.plot(thrs, rates, marker='o'); plt.axhline(0.85, color='r', linestyle='--'); plt.show()
print('mean per dim @ t=8.0:', sd['threshold_8.0']['mean_per_dim'])
```
