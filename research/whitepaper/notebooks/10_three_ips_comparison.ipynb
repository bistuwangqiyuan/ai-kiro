```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
import sys; sys.path.insert(0, str(ROOT.parent.parent))
from research.whitepaper.models import cost_model
scenarios = {
    'bestseller': dict(target_seconds=120, n_characters=8, n_scenes=12, dialogue_chars=4500),
    'average':    dict(target_seconds=90,  n_characters=4, n_scenes=6,  dialogue_chars=3000),
    'flop':       dict(target_seconds=60,  n_characters=3, n_scenes=4,  dialogue_chars=2000),
}
for name, kw in scenarios.items():
    for tier in ('H','M','L'):
        ec = cost_model.per_episode_cost(tier=tier, **kw)
        print(name, tier, '->', round(ec.total_with_retry_cny,2))
```
