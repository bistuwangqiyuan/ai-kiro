# Cost vs Latency vs Quality Pareto frontier


_36 candidate points; non-dominated subset highlighted._

> Reproducible from `data/computed/*.json` written by `run_all.py`.


```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
par = load('pareto')
all_c = par['all_candidates']; front = par['frontier']
plt.scatter([c['cost_cny'] for c in all_c], [c['latency_s']/60 for c in all_c], alpha=0.4)
plt.scatter([c['cost_cny'] for c in front], [c['latency_s']/60 for c in front], color='red', marker='*', s=120)
plt.xlabel('CNY/ep'); plt.ylabel('P95 latency (min)'); plt.title('Pareto frontier'); plt.show()
for f in front: print(f)
```
