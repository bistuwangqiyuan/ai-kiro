# M/M/16 throughput


_Episodes/hour saturation curve and steady-state utilisation._

> Reproducible from `data/computed/*.json` written by `run_all.py`.


```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
thr = load('throughput'); print('eps/hr default c=', thr['episodes_per_hour_at_default_c'])
lambdas = [k for k in thr if k.startswith('lambda_')]
rho = [thr[k]['rho'] for k in lambdas]
wq = [min(thr[k]['wq_seconds_expected_wait'], 3600) for k in lambdas]
vals = [int(k.split('_')[1]) for k in lambdas]
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(vals, rho, marker='o', label='ρ'); ax2 = ax.twinx(); ax2.plot(vals, wq, marker='s', color='orange', label='Wq (s)')
ax.axhline(0.85, color='r', linestyle='--'); ax.set_xlabel('arrivals/hour'); ax.set_title('M/M/c saturation'); plt.tight_layout(); plt.show()
```
