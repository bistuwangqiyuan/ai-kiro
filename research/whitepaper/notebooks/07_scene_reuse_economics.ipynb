# Scene-library reuse marginal savings


_saving_per_ep_cny vs library_size on log-x._

> Reproducible from `data/computed/*.json` written by `run_all.py`.


```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt
ROOT = Path('..').resolve() if Path('..').name == 'whitepaper' else Path('.').resolve().parent
COMPUTED = ROOT / 'data' / 'computed'
FIGURES = ROOT / 'figures'
def load(name): return json.loads((COMPUTED / f'{name}.json').read_text(encoding='utf-8'))
```

```python
sr = load('scene_reuse'); print('library_size_for_90pct =', sr['library_size_for_90pct'])
sizes = [c['library_size'] for c in sr['curve']]
save = [c['saving_per_ep_cny'] for c in sr['curve']]
plt.plot(sizes, save, marker='o'); plt.xscale('log'); plt.xlabel('library size'); plt.ylabel('CNY saved'); plt.show()
```
